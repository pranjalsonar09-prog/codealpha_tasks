```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
print("Labraries imported successfully!")
```

```python
url="https://books.toscrape.com/"
response=requests.get(url)
print("Status Code:",response.status_code)
```

```python
soup=BeautifulSoup(response.text,"html.parser")
print(soup.title.text)
```

```python
books=soup.find_all("article",class_="product_pod")
print("Number of books found:",len(books))
```

```python
first_book=books[0]

title=first_book.h3.a["title"]
price=first_book.find("p",class_="price_color").text.strip()
rating=first_book.find("p",class_="star-rating")["class"][1]
availability=first_book.find("p",class_="instock").text.strip()


print("Title:",title)
print("Price:",price)
print("Rating:",rating)
print("Availability:",availability)
```

```python
books_data=[]

for book in books:
  title=book.h3.a["title"]
  price=book.find("p",class_="price_color").text.strip()
  rating=book.find("p",class_="star-rating")["class"][1]
  availability=book.find("p",class_="instock").text.strip()

  books_data.append({
      "Title":title,
      "Price":price,
      "Rating":rating,
      "Availability":availability
  })

  print("Books extracted:",len(books_data))
```

```python
df=pd.DataFrame(books_data)
print("Dataset created successfully!")
print("Dataset shape:",df.shape)
df.head()
```

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
from urllib.parse import urljoin

url="https://books.toscrape.com"

response=requests.get(url,headers={"User-Agent":"Mozilla/5.0"},timeout=15)
print("Status code:",response.status_code)

soup= BeautifulSoup(response.text,"html.parser")
all_books=soup.select("article.product_pod")
print("Books found:",len(all_books))

data=[]

for book in all_books:

  title=book.select_one("h3 a")["title"]

  price=book.select_one("p.price_color").get_text(strip=True)

  rating=book.select_one("p.star-rating")["class"][1]

  availability = book.select_one("p.instock.availability").get_text(" ",strip=True)

  product_url = urljoin(url,book.select_one("h3 a")["href"]
  )
  data.append([
      title,
      price,
      rating,
      availability,
      product_url
  ])
  df=pd.DataFrame(data,columns=[
      "Title",
      "Price" ,
      "Rating",
      "Availability",
      "Product_url "
      ])
  print("\n+=========================================")
  print("FINAL DATASET")
  print("\n+=========================================")
  print("Books collected:",len(df))
  print("Dataset sahpe:",df.shape)

  df.head()

```

```python
print("====FINAL DATASET CHECK====")
print("Shape:",df.shape)

print("\nColumns:")
print(df.columns.tolist())

print("\nMissing Values:")
print(df.duplicated().sum())

print("\nDuplicate Rows:")
print(df.isnull().sum())

print("\nFirst 5 Records:")
display(df.head())
```

```python
df.to_csv("books_data.csv",index=False)
print("books_data.csv created successfully!")
```
